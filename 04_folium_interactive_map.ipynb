# 4. Folium Interactive Launch-Site Map

```python
import pandas as pd, folium\nfrom folium.plugins import MarkerCluster\ndf=pd.read_csv('data/launches_clean.csv').dropna(subset=['Latitude','Longitude'])\nsites=df.groupby(['LaunchSite','Latitude','Longitude'],as_index=False).agg(launches=('Class','size'),success_rate=('Class','mean'))\nm=folium.Map(location=[sites.Latitude.mean(),sites.Longitude.mean()],zoom_start=4)\ncluster=MarkerCluster().add_to(m)\nfor _,r in sites.iterrows():\n    popup='{}<br>Launches: {}<br>Landing success: {:.1%}'.format(r.LaunchSite,r.launches,r.success_rate)\n    folium.Marker([r.Latitude,r.Longitude],popup=popup).add_to(cluster)\nm.save('data/folium_launch_sites.html')\nm
```
