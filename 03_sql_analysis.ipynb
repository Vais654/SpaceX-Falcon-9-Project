# 3. SQL Analysis with SQLite

```python
import pandas as pd\nfrom sqlalchemy import create_engine\ndf=pd.read_csv('data/launches_clean.csv')\nengine=create_engine('sqlite:///data/spacex.db')\ndf.to_sql('launches',engine,if_exists='replace',index=False)
```

```python
q='SELECT LaunchSite, COUNT(*) AS launches, ROUND(AVG(Class)*100,2) AS success_pct FROM launches GROUP BY LaunchSite ORDER BY success_pct DESC'\ndisplay(pd.read_sql(q,engine))
```

```python
q='SELECT Year, COUNT(*) AS launches, ROUND(AVG(Class)*100,2) AS success_pct FROM launches GROUP BY Year ORDER BY Year'\ndisplay(pd.read_sql(q,engine))
```

```python
q='SELECT Rocket, COUNT(*) AS launches, SUM(Class) AS successful_landings FROM launches GROUP BY Rocket ORDER BY launches DESC'\ndisplay(pd.read_sql(q,engine))
```
